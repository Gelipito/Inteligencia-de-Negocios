```python
import pandas as pd
```

```python
df_energia= pd.read_csv("consumo_energia_iot - consumo_energia_iot.csv",index_col=0)
df_energia
```

```python
#Punto 1.1: Extrae de forma aislada la columna Consumo_kWh como un objeto pd.Series
consumo=df_energia["Consumo_kWh"]
consumo,type(consumo)
```

```python
#Punto 1.2 : y crea una nueva vista con las columnas Municipio, Estrato y Voltaje_Promedio.
nueva_vista=df_energia[["Municipio","Estrato","Voltaje_Promedio"]]
nueva_vista
```

```python
#Punto 2:Mediante el indexador .loc[], extrae todos los datos de la fila ubicada en el índice 150.

row150=df_energia.loc[[150]]
row150
```

```python
#Punto 3:  Detecta y remueve las filas redundantes con .drop_duplicates() para garantizar que cada registro sea único en la tabla.

df_energia.drop_duplicates()
```

```python
#Punto4: Filtra los registros para identificar predios con un consumo elevado, seleccionando aquellos con Consumo_kWh > 250.0.
PrediosConConsumoAlto=df_energia[df_energia["Consumo_kWh"]>250.0]
PrediosConConsumoAlto
```

```python
#Punto 5: Utiliza .isin() para recuperar únicamente los medidores instalados en los municipios de Chinchiná y Villamaría.
filtro = df_energia[
    df_energia["Municipio"].isin(["Chinchiná", "Villamaría"])
]
filtro
```

```python
#Punto 6:Utilizando el método .between(), extrae todos los registros cuyo Voltaje_Promedio se mantenga en el rango óptimo de 118.0 a 122.0 voltios inclusive.
VoltajePromedio=df_energia[df_energia["Voltaje_Promedio"].between(118.0,122.0)]
VoltajePromedio
```

```python

#Punto 7:Aplica una consulta con .query() para filtrar aquellos inmuebles donde Estrato == 'Comercial' y el Estado_Pago == 'Mora'.
queryEnergia= df_energia.query("Estrato=='Comercial' & Estado_Pago == 'Mora'")

queryEnergia
```

```python
#Punto 8: Construye una condición lógica con el operador & para encontrar lecturas donde el Tipo_Medidor comience con "B" (.str.startswith()) y las Horas_Pico_kWh sean mayores a 40.0.
filtro8=df_energia[(df_energia["Tipo_Medidor"].str.startswith("B")) & (df_energia["Horas_Pico_kWh"]>40.0)]
filtro8
```

```python
#9: Tratamiento e Imputación de Nulos

filtro9=df_energia.fillna(
    {
        "Consumo_kWh": df_energia["Consumo_kWh"].mean(),
        "Voltaje_Promedio":120.0
    }
)
filtro9.loc["MED-5010"]
```

```python
#verificar nulos para poder comprobar el resultado en el punto 9 con una fila
cambios = df_energia[
    df_energia["Consumo_kWh"].isna() |
    df_energia["Voltaje_Promedio"].isna()
]

cambios
```

```python
#Punto 10:Transformación Condicional con .apply()

def  nivel_taritario(n):

    if n<120.0:
        return "Bajo"
    elif n<= 250.0:
        return"Medio"
    else:
        return "Alto"



df_energia["Nivel_Consumo"]= df_energia["Consumo_kWh"].apply(nivel_taritario)
df_energia

```

```python
#11.Renombrado y Reordenamiento de Columnas

df_energia.rename(
    columns={
        "Consumo_kWh": "Energia_kWh",
        "Tipo_Medidor": "Tecnologia"
    },inplace=True
)

cols = ["Nivel_Consumo"]

df_energia = df_energia[
    cols + [col for col in df_energia.columns if col not in cols]
]

df_energia
```

```python
groupsbyCity = df_energia.groupby("Municipio")




agregate_data=groupsbyCity.agg(
{
    "Energia_kWh":"mean",
    "Horas_Pico_kWh":"sum"
}
)

agregate_data
```

```python
#11

def dispersion(series):
    return series.max()-series.min()


groupByEstrato=df_energia.groupby("Estrato").agg({
    "Energia_kWh":dispersion,
    "Voltaje_Promedio": dispersion

})
groupByEstrato
```

```python


for estrato, grupo in df_energia.groupby("Estrato"):

    energia: pd.Series = grupo["Energia_kWh"]

    print(
        estrato,
        "min =", energia.min(),
        "max =", energia.max(),
        "range =", energia.max() - energia.min()
    )
```

```python
#12 Agrupamiento Multinivel:

groupMulti=df_energia.groupby(["Municipio","Estrato"])

groupMulti.agg(
    {
        "Energia_kWh":"mean",
        "Voltaje_Promedio":"mean"
    }
)
```

```python
#Viendo como se visualizan los datos para hacer la alternativa del punto 12
h=df_energia.groupby(["Municipio","Estrato"])
h.groups
```

```python
#Otra forma forma de hacer el punto 12

for  dupla,datos in df_energia.groupby(["Municipio","Estrato"]):

    Municipio=dupla[0]
    estrato=dupla[1]

    Promedio_Energia_kWh= datos["Energia_kWh"].mean()
    Promedio_Voltaje_Promedio=datos["Voltaje_Promedio"].mean()

    

    print(
        (Municipio,estrato),
        "Promedio energia = " , Promedio_Energia_kWh,
        "Promedio Voltaje = " , Promedio_Voltaje_Promedio
        
    )
```
